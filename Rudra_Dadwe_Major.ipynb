```python
{
  "nbformat": 4,
  "nbformat_minor": 0,
  "metadata": {
    "colab": {
      "provenance": []
    },
    "kernelspec": {
      "name": "python3",
      "display_name": "Python 3"
    },
    "language_info": {
      "name": "python"
    }
  },
  "cells": [
    {
      "cell_type": "code",
      "source": [
        "# ================================\n",
        "# Student Performance Progression & Risk Analyzer\n",
        "# ================================\n",
        "\n",
        "import numpy as np\n",
        "import pandas as pd\n",
        "import matplotlib.pyplot as plt\n",
        "import seaborn as sns\n",
        "\n",
        "sns.set(style=\"whitegrid\")\n",
        "plt.rcParams[\"figure.figsize\"] = (8,5)\n",
        "\n",
        "# ----------------\n",
        "# Load Data\n",
        "# ----------------\n",
        "df = pd.read_csv(\"/content/sample_data/StudentsPerformance.csv\")\n",
        "\n",
        "print(\"Shape:\", df.shape)\n",
        "print(\"\\nColumns:\\n\", df.columns)\n",
        "print(\"\\nInfo:\")\n",
        "df.info()\n",
        "print(\"\\nDescribe:\")\n",
        "print(df.describe())\n",
        "\n",
        "# ----------------\n",
        "# Data Cleaning\n",
        "# ----------------\n",
        "df.columns = (\n",
        "    df.columns\n",
        "    .str.lower()\n",
        "    .str.replace(\" \", \"_\")\n",
        "    .str.replace(\"/\", \"_\")\n",
        ")\n",
        "\n",
        "print(\"\\nMissing values:\\n\", df.isna().sum())\n",
        "\n",
        "df[\"average_score\"] = df[[\"math_score\",\"reading_score\",\"writing_score\"]].mean(axis=1)\n",
        "\n",
        "# ----------------\n",
        "# Simulated Performance Progression\n",
        "# ----------------\n",
        "df[\"delta_math_reading\"] = df[\"reading_score\"] - df[\"math_score\"]\n",
        "df[\"delta_reading_writing\"] = df[\"writing_score\"] - df[\"reading_score\"]\n",
        "\n",
        "def performance_trend(row):\n",
        "    total_change = row[\"writing_score\"] - row[\"math_score\"]\n",
        "    if total_change > 5:\n",
        "        return \"Improving\"\n",
        "    elif total_change < -5:\n",
        "        return \"Declining\"\n",
        "    else:\n",
        "        return \"Stable\"\n",
        "\n",
        "df[\"performance_trend\"] = df.apply(performance_trend, axis=1)\n",
        "\n",
        "# ----------------\n",
        "# Consistency Score\n",
        "# ----------------\n",
        "df[\"score_std\"] = df[[\"math_score\",\"reading_score\",\"writing_score\"]].std(axis=1)\n",
        "df[\"consistency_score\"] = 1 - (df[\"score_std\"] / df[\"score_std\"].max())\n",
        "\n",
        "# ----------------\n",
        "# Subject Difficulty Index\n",
        "# ----------------\n",
        "subject_means = df[[\"math_score\",\"reading_score\",\"writing_score\"]].mean()\n",
        "difficulty_index = 1 - (subject_means / 100)\n",
        "\n",
        "difficulty_df = difficulty_index.reset_index()\n",
        "difficulty_df.columns = [\"subject\",\"difficulty_index\"]\n",
        "difficulty_df = difficulty_df.sort_values(\"difficulty_index\", ascending=False)\n",
        "\n",
        "print(\"\\nSubject Difficulty Index:\\n\", difficulty_df)\n",
        "\n",
        "# ----------------\n",
        "# Risk Classification\n",
        "# ----------------\n",
        "def risk_level(row):\n",
        "    if row[\"average_score\"] < 60 and row[\"consistency_score\"] < 0.5:\n",
        "        return \"High Risk\"\n",
        "    elif row[\"average_score\"] < 75:\n",
        "        return \"Medium Risk\"\n",
        "    else:\n",
        "        return \"Low Risk\"\n",
        "\n",
        "df[\"risk_level\"] = df.apply(risk_level, axis=1)\n",
        "\n",
        "# ----------------\n",
        "# Engagement Proxy (Test Prep)\n",
        "# ----------------\n",
        "prep_analysis = df.groupby(\"test_preparation_course\")[\"average_score\"].mean()\n",
        "print(\"\\nAverage Score by Test Preparation:\\n\", prep_analysis)\n",
        "\n",
        "# ----------------\n",
        "# Visualizations\n",
        "# ----------------\n",
        "sns.histplot(df[\"average_score\"], bins=20, kde=True)\n",
        "plt.title(\"Distribution of Average Scores\")\n",
        "plt.show()\n",
        "\n",
        "sns.boxplot(data=df[[\"math_score\",\"reading_score\",\"writing_score\"]])\n",
        "plt.title(\"Subject-wise Score Distribution\")\n",
        "plt.show()\n",
        "\n",
        "sns.barplot(data=difficulty_df, x=\"subject\", y=\"difficulty_index\")\n",
        "plt.title(\"Subject Difficulty Index\")\n",
        "plt.show()\n",
        "\n",
        "sns.countplot(x=\"risk_level\", data=df)\n",
        "plt.title(\"Student Risk Level Distribution\")\n",
        "plt.show()\n",
        "\n",
        "sns.barplot(x=\"test_preparation_course\", y=\"average_score\", data=df)\n",
        "plt.title(\"Impact of Test Preparation on Performance\")\n",
        "plt.show()\n",
        "\n",
        "# ----------------\n",
        "# Insights\n",
        "# ----------------\n",
        "print(\"\\nKEY INSIGHTS:\")\n",
        "print(\"1. Test preparation completion is associated with higher average scores.\")\n",
        "print(\"2. Math appears as the most difficult subject on average.\")\n",
        "print(\"3. High-risk students show both low performance and instability.\")\n",
        "print(\"4. Consistency is a strong indicator of academic reliability.\")\n",
        "print(\"5. Many medium-risk students are stable but underperforming.\")\n",
        "print(\"6. Subject score variance indicates curriculum difficulty imbalance.\")\n"
      ],
      "metadata": {
        "id": "3AgRItB4ELBb"
      },
      "execution_count": null,
      "outputs": []
    },
    {
      "cell_type": "code",
      "source": [],
      "metadata": {
        "id": "S_6HrWsAFGi1"
      },
      "execution_count": null,
      "outputs": []
    }
  ]
}
```
